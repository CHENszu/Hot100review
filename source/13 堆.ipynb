# 13 堆

## 13.1 数组中的第K个最大元素


```python
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        import heapq        # 用Python自带堆库
        heap = []           # 创建一个空的小顶堆
        
        # 遍历数组里每一个数字
        for num in nums:
            heapq.heappush(heap, num)   # 把数字塞进堆
            
            # 重点！！！
            if len(heap) > k:          # 如果堆里的数字超过 k 个
                heapq.heappop(heap)    # 就把最小的那个扔掉！
        
        # 最后堆里只剩下 k 个最大的数
        # 堆顶就是第 k 大
        return heap[0]
```

## 13.2 前K个高频元素

```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        import heapq
        from collections import defaultdict
        # 统计频次
        count = defaultdict(int)
        for num in nums:
            count[num] += 1
        
        heap = []
        for num, freq in count.items():
            heapq.heappush(heap, (freq, num))
            # 堆超过k个，弹出最小的
            if len(heap) > k:
                heapq.heappop(heap)
        
        # 取出堆中所有数字
        return [item[1] for item in heap]
```
