## 1 数组中的第K个最大元素

给定整数数组 nums 和整数 k，请返回数组中第 k 个最大的元素。  
请注意，你需要找的是数组排序后的第 k 个最大的元素，而不是第 k 个不同的元素。  
你必须设计并实现时间复杂度为 O(n) 的算法解决此问题。

```python
import random

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        target = len(nums) - k
        left,right = 0,len(nums) - 1

        while left <= right:
            mid = nums[random.randint(left,right)]
            l,i,r = left,left,right
            while i <= r:
                if nums[i] < mid:
                    nums[l],nums[i] = nums[i],nums[l]
                    l += 1
                    i += 1
                elif nums[i] == mid:
                    i += 1
                else:
                    nums[i],nums[r] = nums[r],nums[i]
                    r -= 1

            if target < l:
                right = l -1
            elif target > r:
                left = r + 1
            else:
                return nums[target]
```

## 2 合并两个有序数组

```python
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        # 三个指针初始位置
        p1 = m - 1
        p2 = n - 1
        p = m + n - 1

        # 从后往前遍历，只要还有元素没比较就继续
        while p1 >= 0 and p2 >= 0:
            if nums1[p1] > nums2[p2]:
                # 大的元素放到最后
                nums1[p] = nums1[p1]
                p1 -= 1
            else:
                nums1[p] = nums2[p2]
                p2 -= 1
            p -= 1

        # 剩余：如果 nums2 还有元素没放完，直接覆盖到 nums1 前面
        nums1[:p2 + 1] = nums2[:p2 + 1]
```

## 3 冒泡排序

```python
def sort(nums):
    n = len(nums)
    for i in range(n):
        flag = 0  # 标记本轮是否交换
        # 遍历未排序区间
        for j in range(n-i-1):
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]
                flag = 1
        # 无交换说明已有序，直接跳出
        if not flag:
            break
    return nums
```
